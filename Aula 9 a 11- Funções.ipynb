# O que são funções? 

Em Python, funções são trechos de código nomeados que executam operações definidas e podem ser usados quantas vezes você quiser. Em vez de repetir o mesmo conjunto de instruções várias vezes, você “empacota” essa lógica em uma função e a chama sempre que precisar.

Por que usar funções?
- Clareza e organização: dividem programas complexos em partes menores, facilitando a leitura e a manutenção.
- Reaproveitamento: concentram trechos de lógica em um só lugar, evitando cópias espalhadas pelo código.
- Ocultamento de detalhes: você foca no que a função faz sem se preocupar com a forma como ela faz.
- Estrutura modular: promovem a independência entre diferentes blocos de código, o que simplifica testes e atualizações.



### Como criar uma função em Python

Use a palavra-chave def, seguida pelo nome da função e parênteses (que podem receber parâmetros), e finalize com dois pontos. Todo o código que pertence à função deve ter um nível de indentação a mais:

```python
def saudacao():
    print("Olá, seja bem-vindo a Universidade dos Dados")
```

```python
saudacao()
```

```python
saudacao()
```

### Como utilizar parâmetros na função

```python
def soma(a, b):
    soma = a + b
    print(f"A soma de {a} + {b} é {soma}")
```

```python
soma(10, 5)
```

```python
soma(3, 2)
```

```python
def multiplicacao(a, b, c, d):
    multiplicacao = a * b * c * d
    print(f"A multiplicação de {a}, {b}, {c} e {d} é {multiplicacao}")
```

```python
multiplicacao(a = 10, b = 5, c = 3, d = 1)
```

```python
multiplicacao(10, 5, 3, 1)
```

```python
def conta(a, b, c, d):
    conta = a * b
    conta = conta - c + d
    print(f"A conta com {a}, {b}, {c} e {d} é {conta}")
```

```python
conta(2, 2, 4, 5)
```

### Argumentos Não Nomeados

```python
def soma(*valores):
    return sum(valores)
```

```python
soma(1,2,3,4,5)
```

```python
def soma(*args):
    resultado = 0

    for arg in args:
        resultado = resultado + arg
        
    return resultado
```

```python
soma(1,2,3,4,5)
```

```python
def conta(arg1=1, arg2=1, *args):
    
    resultado = 0

    for arg in args:
        resultado = resultado + arg

    resultado = resultado * arg1 * arg2
        
    return resultado
```

```python
conta(2, 3, 1, 2, 3, 4, 5)
```

### Argumento Nomeado

```python
def argumento_nomeado(**kwargs):
    print(kwargs)
```

```python
argumento_nomeado(nome="Yukio", idade=38, estado="SP")
```

```python
def argumento_nomeado(**kwargs):
    for chave, valor in kwargs.items():
        print(f"O campo {chave} recebeu o valor {valor}.")
```

```python
argumento_nomeado(nome="Yukio", idade=38, estado="SP")
```
