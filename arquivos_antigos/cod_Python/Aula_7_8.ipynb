## Loop

For

```python
for i in [1,2,3,4,5]: 
    print(i)
```

```python
for i in [10,20,30,40,50]:
    valor = i**2
    print(valor)
```

```python
notas = [7, 8, 9.5, 7.6, 8, 4, 2]

aprovados = 0
reprovados = 0

for nota in notas:
    if nota < 6:
        reprovados +=1
    else:
        aprovados +=1

print(f"Temos {reprovados} alunos reprovados e {aprovados} alunos aprovados.")
```

While

```python
contador = 1

while contador < 5:
    print(contador)
    contador += 1
    print(contador)
```

```python
opcao = 0

while opcao != "3":
    print("\n=== Menu ===")
    print("1 -  Ver Mensagem")
    print("2 - Ver Outra Mensagem")
    print("3 - Sair")

    opcao = input("Escolha uma opção:")

    if opcao == "1":
        print('Opção 1')
    elif opcao == "2":
        print("Opção 2")
    elif opcao == "3":
        print("Encerrando")
    else:
        print("Op inválida")
```

```python
while True:
    senha = input("Digite a senha:")
    if senha == "1234":
        print("Liberado")
        break
    else:
        print("Senha incorreta.")
```

```python
tentativas_max = 3
tentativas_realizadas = 0
senha_correta = "1234"

while tentativas_realizadas < tentativas_max:
    senha_digitada = input("Digite sua senha:")
    tentativas_realizadas += 1

    if senha_digitada == senha_correta:
        print("Liberado")
        break
    else:
        print(f"Senha incorreta. Alternativas restantes: {tentativas_max - tentativas_realizadas}")

    if tentativas_realizadas == tentativas_max:
        print("Bloqueado")
```
