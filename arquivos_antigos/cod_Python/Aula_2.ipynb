#### Operadores lógicos


Operador 'and'

```python
idade = int(input('Digite sua idade:'))
status_em = input('Ensino médio concluído ou interrompido')

if idade >= 18 and status_em == 'concluído':
    print("Cadastro aprovado")
else:
    print("Cadastro reprovado")
```

Operador 'or'

```python
anos_cliente = int(input('Há quantos anos é cliente: '))
saldo = int(input('Valor investido no banco: '))

if anos_cliente >= 5 or saldo >= 50000:
    print('Cliente vip!')
else:
    print('Cliente não vip!')
```

Operador 'not'

```python
op = int(input('Selecione uma das opções de 1 a 5: '))

if not op == 3:
    print('Encaminhando para atendente...')
else:
    print('Necessário falar com gerente da sua conta.')
```
