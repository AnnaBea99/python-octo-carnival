# Exercício de Filtros

```python
import pandas as pd

df = pd.read_csv('./arquivos/insurance.csv')
```

1. Quantas pessoas possuem filhos?



```python
df.loc[df['children'] > 0].shape[0]
```

2. Selecione as linhas da posição 100 até a 120, mantendo apenas as colunas `age`, `bmi` e `charges`.



```python
df.loc[100:120, ['age', 'bmi', 'charges']]
```

3. Quantas mulheres estão na região `southeast`?



```python
df.loc[(df['sex'] == 'female') & (df['region'] == 'southeast')].shape[0]
```

4. Selecione apenas os homens com IMC (`bmi`) acima de 30.



```python
df.loc[
    (df['sex'] == 'male')
    &
    (df['bmi'] > 30)
]
```

5. Use `query` para filtrar pessoas com `age > 40` e `charges > 20000`.



```python
df.query("age > 40 and charges > 20000")
```

6. Filtre com `loc` todas as pessoas fumantes com mais de 30 anos e que têm pelo menos 1 filho.



```python
df.loc[
    (df['smoker'] == 'yes')
    &
    (df['age'] > 30)
    &
    (df['children'] > 0)
]
```

7. Usando `query`, selecione pessoas da região `southwest` que não são fumantes e têm IMC entre 25 e 35.



```python
df.query("region == 'southwest' and smoker == 'no' and 25 <= bmi <= 35")
```

8. Filtre todos os registros onde o número de filhos é par e o valor de `charges` está entre 10.000 e 20.000.



```python
df.query("children % 2 == 0 and 10000 <= charges <= 20000")
```

```python
df.loc[(df['children'] % 2 == 0)
       &
       (df['charges'].between(10000, 20000))
]
```

9. Use `iloc` para obter as últimas 10 linhas e as 4 primeiras colunas.



```python
df.iloc[-10:, :4]
```

10. Usando `query`, filtre pessoas com mais de 3 filhos, que sejam fumantes, e estejam na região `northeast` ou `northwest`.

```python
df.query("children > 3 and smoker == 'yes' and (region == 'northeast' or  region == 'northwest')")
```
