Você tem os seguintes dados de custos mensais de dois departamentos de uma empresa:

* `custo_departamento_A = 12000`

Existe uma taxa fixa de manutenção de 5% aplicada ao custo total dos departamentos. Sabendo disso:

1. Calcule o custo total somando `custo_departamento_A` e `custo_departamento_B`.
2. Calcule o valor da taxa de manutenção (5% do custo total).
3. Armazene o resultado final (custo total mais taxa) em uma variável `custo_com_taxa`.

**Resposta:**

```python
custo_departamento_A = 12000
custo_departamento_B = 8000

# 1. Custo total
custo_total = custo_departamento_A + custo_departamento_B  # 20000

# 2. Valor da taxa de manutenção (5% do custo_total)
taxa_manutencao = custo_total * 0.05  # 1000

# 3. Resultado final (custo_total + taxa)
custo_com_taxa = custo_total + taxa_manutencao  # 21000

print("Custo total:", custo_total)
print("Taxa de manutenção (5%):", taxa_manutencao)
print("Custo com taxa:", custo_com_taxa)
```

**Exercício 4:**
Uma empresa quer saber o lucro operacional anual. Os dados disponíveis são:

* `receita_anual = 250000`
* `custo_operacional_mensal = 15000`
* `despesa_variavel_mensal = 5000`
* `depreciacao_anual = 12000`

Sabendo disso:

1. Calcule a despesa operacional anual somando `custo_operacional_mensal * 12` e `despesa_variavel_mensal * 12`.
2. Subtraia da `receita_anual` a `despesa operacional anual` e a `depreciacao_anual`.
3. Armazene o resultado em `lucro_operacional_anual`.

**Resposta:**

```python
receita_anual = 250000
custo_operacional_mensal = 15000
despesa_variavel_mensal = 5000
depreciacao_anual = 12000

# 1. Despesa operacional anual
despesa_operacional_anual = (custo_operacional_mensal * 12) + (despesa_variavel_mensal * 12)
# 180000 + 60000 = 240000

# 2. Lucro operacional anual
lucro_operacional_anual = receita_anual - despesa_operacional_anual - depreciacao_anual
# 250000 - 240000 - 12000 = -2000

print("Despesa operacional anual:", despesa_operacional_anual)
print("Lucro operacional anual:", lucro_operacional_anual)
```

**Exercício 5:**
Um analista precisa calcular o indicador EBITDA (Earnings Before Interest, Taxes, Depreciation and Amortization) para um conjunto de dados simplificado. As informações são:

* `receita_trimestral = 800000`
* `custo_produtos = 300000`
* `despesa_operacional = 100000`
* `despesa_financeira = 20000`
* `impostos = 50000`
* `depreciacao = 30000`
* `amortizacao = 10000`

1. Calcule o lucro operacional bruto (`receita_trimestral - custo_produtos - despesa_operacional`).
2. Subtraia `despesa_financeira` e `impostos` desse lucro operacional bruto para obter o lucro líquido antes de depreciação e amortização.
3. Por fim, calcule o EBITDA somando `depreciacao` e `amortizacao` ao valor obtido no passo anterior. Armazene em `ebitda_trimestral`.

**Resposta:**

```python
receita_trimestral = 800000
custo_produtos = 300000
despesa_operacional = 100000
despesa_financeira = 20000
impostos = 50000
depreciacao = 30000
amortizacao = 10000

# 1. Lucro operacional bruto
lucro_operacional_bruto = receita_trimestral - custo_produtos - despesa_operacional
# 400000

# 2. Lucro líquido antes de depreciação e amortização
lucro_liquido_antes_depre_e_amort = lucro_operacional_bruto - despesa_financeira - impostos
# 330000

# 3. EBITDA trimestral
ebitda_trimestral = lucro_liquido_antes_depre_e_amort + depreciacao + amortizacao
# 370000

print("Lucro operacional bruto:", lucro_operacional_bruto)
print("Lucro líquido antes de depreciação e amortização:", lucro_liquido_antes_depre_e_amort)
print("EBITDA trimestral:", ebitda_trimestral)
```

**Exercício 5:**
Peça ao usuário que informe as três notas em uma única linha, separadas por vírgula (por exemplo: `7.5, 8.0, 6.0`).

1. Leia essa string com `input()` e use `split(",")` para obter cada nota.
2. Converta as três partes em `float` e calcule a média aritmética.
3. Verifique, com um operador booleano, se a média é maior que 6, armazenando o resultado em uma variável chamada `aprovado` (True ou False).
4. Por fim, exiba duas informações usando `print()` e f-string:

   * O valor de `aprovado` (True/False).
   * A média calculada.
     Exemplo de saída esperada (caso a média seja 7.0):

```
Aprovado: True
Média: 7.0
```

**Resposta**:

```python
# Leitura das três notas em uma única linha
notas = input("Digite as 3 notas separadas por vírgula (por exemplo: 7.5, 8.0, 6.0): ")

# Separar as notas e converter para float
lista_notas = notas.split(",")
nota1 = float(lista_notas[0].strip())
nota2 = float(lista_notas[1].strip())
nota3 = float(lista_notas[2].strip())

# Cálculo da média
media = (nota1 + nota2 + nota3) / 3

# Verificar se a média é maior que 6
aprovado = media > 6

# Exibir resultado booleano e a média usando f-string
print(f"Aprovado: {aprovado}")
print(f"Média: {media}")
```

**Exercício 5:**
Você recebeu um dicionário chamado `dados_vendas` que contém, para cada mês do ano, o valor total de vendas. Por exemplo:

```python
dados_vendas = {
    "janeiro": 12000.0,
    "fevereiro": 15000.0,
    "março": 13000.0,
    "abril": 14000.0,
    "maio": 16000.0
}
```

1. Adicione ao dicionário o mês de `junho` com valor de 17000.0.
2. Calcule o total anual (soma de todos os valores presentes no dicionário).
3. Identifique mês com maior venda e mês com menor venda, armazenando os nomes desses meses em `mes_maior_venda` e `mes_menor_venda`.
4. Crie uma lista chamada `meses` que contenha apenas as chaves (meses) do dicionário, e uma lista chamada `valores` que contenha apenas os valores de vendas.
5. Exiba todos esses resultados (total anual, `mes_maior_venda`, `mes_menor_venda`, `meses`, `valores`) usando prints adequados.

**Resposta:**

```python
dados_vendas = {
    "janeiro": 12000.0,
    "fevereiro": 15000.0,
    "março": 13000.0,
    "abril": 14000.0,
    "maio": 16000.0
}

# 1. Adicionar junho
dados_vendas["junho"] = 17000.0

# 2. Calcular total anual
total_anual = sum(dados_vendas.values())

# 3. Identificar mês com maior e menor venda
mes_maior_venda = max(dados_vendas, key=dados_vendas.get)
mes_menor_venda = min(dados_vendas, key=dados_vendas.get)

# 4. Criar lista de meses (chaves) e lista de valores
meses = list(dados_vendas.keys())
valores = list(dados_vendas.values())

# 5. Exibindo resultados
print(f"Total anual de vendas: {total_anual}")
print(f"Mês com maior venda: {mes_maior_venda} ({dados_vendas[mes_maior_venda]})")
print(f"Mês com menor venda: {mes_menor_venda} ({dados_vendas[mes_menor_venda]})")
print(f"Lista de meses: {meses}")
print(f"Lista de valores: {valores}")
```
