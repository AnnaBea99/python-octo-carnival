# Aula: Loop em Python

Na programação, um loop serve para repetir um bloco de código várias vezes. Isso é muito útil quando precisamos realizar a mesma tarefa com vários dados, como verificar notas de alunos, somar números, processar arquivos, entre outros.

Existem dois principais tipos de loop em Python: for e while.

## Primeira Parte: Loop com for

```python
for i in [1,2,3,4,5]:
    print(i)
```

```python
for i in [10,20,30,40,50]:
    valor = i**2
    print(valor)
```

```python
notas = [5, 6, 1, 3.5, 6.7, 9.6, 10, 8, 7.5, 1.3]
```

```python
aprovados = 0
reprovados = 0

for nota in notas:
    print("Nota do aluno: ", nota)
    if nota < 6:
        reprovados = reprovados + 1
        print("Reprovado!")
    else:
        aprovados = aprovados + 1
        print("Aprovado")
```

```python
print(f"Temos {reprovados} alunos reprovados e {aprovados} alunos aprovados!")
```

**Dica:** Você pode usar aprovados += 1 ao invés de aprovados = aprovados + 1

## Segunda Parte: Loop com while

```python
valor = 0

for i in [10,15,30,35,50]:
    print("i = ", i)
    if i%2==0:
        valor += 1
    else:
        valor = valor
    print("valor = ", valor)
```

```python
for n in range(5):
    print(n)
```

```python
contador = 0
for n in range(5):
    contador += 1

print(contador)
```

```python
contador = 1

while contador < 5:
    contador += 1
    print("depois da ação", contador)
```

```python
opcao = ""

while opcao != "3":
    print("\n=== MENU ===")
    print("1 - Ver Mensagem")
    print("2 - Ver Outra Mensagem")
    print("3 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        print("Opção 1 selecionada")
    elif opcao == "2":
        print("Opção 2 selecionada")
    elif opcao == "3":
        print("Encerrando o programa...")
    else:
        print("Opção inválida. Tente novamente")
```

```python
while True:
    senha = input("Digite a senha: ")
    if senha == "1234":
        print("Acesso liberado!")
        break
    else:
        print("Senha incorreta. Tente novamente.")
```

```python
tentativas_maximas = 3
tentativas_realizadas = 0
senha_correta = "senha123"

while tentativas_realizadas < tentativas_maximas:
    senha_digitada = input("Digite sua senha: ")
    tentativas_realizadas += 1

    if senha_digitada == senha_correta:
        print("Acesso concedido!")
        break  # Sai do loop assim que a senha estiver correta
    else:
        print(f"Senha incorreta. Você tem mais {tentativas_maximas - tentativas_realizadas} tentativas.")

    if tentativas_realizadas == tentativas_maximas:
        print("Número máximo de tentativas excedido. Acesso bloqueado.")

```
