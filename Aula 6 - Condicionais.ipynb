# Aula: Condicionais em Python


Em programação, **condicionais** permitem que o código **tome decisões**. Com base em uma condição (como idade, tempo, valor, etc.), o programa pode executar **um caminho diferente**.

Em Python, usamos as palavras-chave `if`, `elif` (abreviação de "else if") e `else` para controlar o fluxo do programa com base nessas condições.


Veja que a tradução de **if** é "se", logo, estamos dizendo uqe se algo acontece, o programa deve realizar uma certa tarefa.
Também sabemos que **else** significa "do contrário", este é o comando que vai indicar o que fazer caso a condição não seja atendida!

```python
nota = 5

if nota < 6:
    print("Reprovado!")
else:
    print("Aprovado")
```

Imagine que você tenha 3 situações:
- Se o aluno tirar mais que 8, a nota foi ótima!
- Se o aluno tirar maior ou igual a 6 e menor do que 8, a nota foi boa.
- Caso contrário, a nota foi ruim.

```python
if nota > 8:
    print("Sua nota foi ótima!")
elif nota >= 6 and nota < 8:
    print("Sua nota foi boa!")
else: 
    print("Sua nota foi ruim!")
```

```python
nota = 9.5

if nota > 8:
    print("Sua nota foi ótima!")
elif nota >= 6 and nota < 8:
    print("Sua nota foi boa!")
else: 
    print("Sua nota foi ruim!")
```

```python
nota = float(input("Insira sua nota: "))
if nota > 8:
    print("Sua nota foi ótima!")
elif nota >= 6 and nota < 8:
    print("Sua nota foi boa!")
else: 
    print("Sua nota foi ruim!")
```

```python
prazo = int(input("Qual o prazo da sua aplicação?"))
rendimento = float(input("Qual o rendimento da sua aplicação?"))

if prazo < 180:
    aliquota = 0.225
elif prazo <= 360:
    aliquota = 0.20
elif prazo <= 720:
    aliquota = 0.175
else:
    aliquota = 0.15

print(f"O valor recolhido de imposto é: {aliquota * rendimento}")
```
