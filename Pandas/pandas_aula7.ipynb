**Atividade 3:** O time financeiro está elaborando um relatório para entender se o tabagismo está associado a maiores gastos com saúde entre os clientes. Eles te pediram ajuda para gerar os seguintes dados:
- Calcule a média de gastos com saúde dos clientes fumantes. Dica: Para obter a média de uma coluna, você usará o código dataframe.nome_coluna.mean(). Este é o único método que não havia explicado e que você precisará para entregar esta demanda.
- Compare os dois valores e registre, em poucas linhas, se os resultados fazem sentido para você. Se possível, indique uma hipótese que possa justificar a diferença ou ausência dela.

Essas informações serão utilizadas em uma reunião com o setor de produto, que está avaliando políticas diferenciadas para clientes com hábitos de risco.

```python
average_charge_smokers = insurance.query("smoker == 'yes'").charges.mean()
print(f"Média dos gastos dos cliente fumantes: {round(average_charge_smokers)}")
```

```python
average_charge_no_smokers = insurance.query("smoker == 'no'").charges.mean()
print(f"Média dos gastos dos cliente fumantes: {round(average_charge_no_smokers)}")
```

```python
# Representa
rep = average_charge_smokers/average_charge_no_smokers
print(rep)
```

Segundo análises de ambas as médias pessoas fumantes gastam bem mais com saúde do que as não fumantes.
