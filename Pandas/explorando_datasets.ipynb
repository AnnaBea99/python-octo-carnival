### Filtros com Pandas

```python
# pode inserir uma lógica dentro do loc para retornar resultados
df.loc[df['age'] >= 63]
```

```python
df.loc[df['smoker'] == 'yes'].shape
```

```python
df.loc[
    (df['smoker'] == 'yes')
    &
    (df['charges'] < 14000)
    &
    (df['region'] == 'northeast')
]
```

```python
df.loc[
    (df['smoker'] == 'yes')
    |
    (df['sex'] == 'male')
]
```

```python
df.query("age > 50")
```

```python
df.query("region == 'northeast'").shape[0]
```

```python
df.query("smoker == 'yes' and age > 40")
```
