## Condicionais


```python
nota = 5

if nota < 6:
    print("Reprovado")
else:
    print("Aprovado")
```

```python
nota = 9.5

if nota > 8:
    print("Nota ótima")
elif nota >=6 and nota < 8:
    print("Boa nota")
else:
    print("Nota ruim")
```

```python
nota = float(input("Insira sua nota:"))

if nota > 8:
    print("Nota ótima")
elif nota >=6 and nota < 8:
    print("Boa nota")
else:
    print("Nota ruim")
```

```python
prazo = int(input("Qual o prazo da sua aplicação?"))
rendimento = float(input("Qual o rendimento da sua aplicação?"))

if prazo < 180:
    aliquota = 0.225
elif prazo <= 360:
    aliquota = 0.20
elif prazo <= 720:
    aliquota = 0.175
else:
    aliquota = 0.15

print(f"O valor recolhido de imposto é: {aliquota * rendimento}")
```
