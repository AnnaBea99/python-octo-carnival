Por fim, como formatar decimais e valores monetários?

```python
num_socios = 3
participacao = 100/num_socios
print(f"Cada sócio receberá {participacao}")
```

```python
print(f"Cada sócio receberá {participacao:.2f}")
```

```python
salario = 10400
print(f"O salário do {nome} é de R$ {salario:.2f} por mês!")
```

Adicionando separador de milhar:

```python
print(f"O salário de {nome} é de R$ {salario:,.2f} por mês!")
```

Se quiser usar separadores brasileiros:

```python
import locale

locale.setlocale(locale.LC_ALL, 'pt_BR')    # ou 'pt_BR' no Windows

valor_fmt = locale.format_string('%.2f', salario, grouping=True)
print(f"O salário da {nome} é de R$ {valor_fmt} por mês!")
```
