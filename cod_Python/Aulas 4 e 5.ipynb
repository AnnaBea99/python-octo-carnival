# Tipagem Dinâmica vs Estática

## Tipagem Dinâmica

A **tipagem dinâmica** é um jeito de programar onde o **tipo de informação que uma variável guarda é descoberto só na hora que o programa está rodando**, e não antes. Isso quer dizer que você não precisa dizer qual tipo de informação a variável vai guardar quando você a cria. E o tipo pode até mudar enquanto o programa funciona, dependendo do que você coloca dentro dela.

**O que acontece na tipagem dinâmica:**

* **Você não precisa dizer o tipo da variável:** Você só coloca um valor dentro dela, e o computador entende o tipo sozinho.
* **O tipo está ligado ao valor, não à variável:** Uma variável pode guardar números em um momento e texto em outro.
* **Os erros de tipo aparecem quando o programa está rodando:** Se você tentar fazer uma operação errada (como somar um número com uma letra), o erro só vai aparecer quando essa parte do programa for executada.
* **Mais liberdade e para começar rápido:** Como você não precisa se preocupar com os tipos no início, programar pode ser mais rápido e flexível.

**O Python usa tipagem dinâmica.**

**Outras linguagens que também funcionam assim:**

* JavaScript
* Ruby
* PHP
* Perl

## Tipagem Estática

O outro jeito de trabalhar com tipos em programação é a **tipagem estática**. Aqui, **o tipo de informação que uma variável vai guardar é dito claramente pelo programador (ou o computador descobre sozinho antes de rodar o programa)** e esse tipo não muda. A verificação se os tipos estão corretos acontece **antes do programa começar a rodar**.

**O que acontece na tipagem estática:**

* **Geralmente, você precisa dizer o tipo da variável:** Ao criar a variável, você já fala se ela vai guardar números, texto, etc.
* **O tipo fica ligado à variável:** Depois que você define o tipo, a variável só pode guardar informações daquele tipo.
* **Os erros de tipo são encontrados antes do programa rodar:** Se você tentar colocar um tipo de informação errado na variável, o computador avisa antes de executar o programa.
* **Mais segurança e talvez um programa mais rápido:** Encontrar erros antes e ter a informação dos tipos ajuda a criar programas mais seguros e que podem funcionar mais rápido.

**Linguagens que usam tipagem estática:**

* Java
* C
* C++
* C#
* Go

É bom saber que algumas linguagens podem ter um pouco dos dois jeitos (como o Python, que permite dar "dicas" de tipo para ajudar a encontrar erros, mas continua sendo dinâmico na hora de rodar). Mas a forma principal como o Python lida com tipos é a dinâmica.

## 2.6 Dicionários

Dicionários são para guardar pares de informação no formato chave-valor. É bom para associar informações!

```python
cliente = {
    "nome": "Yukio",
    "idade": 37,
    "cidade": "São Paulo"
}
```

```python
print(cliente["cidade"])
```

```python
cliente["idade"] = 38
```

```python

cliente["idade"]
```

```python
type(cliente)
```
